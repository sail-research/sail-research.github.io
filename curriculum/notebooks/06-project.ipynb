# From a broad interest to a testable research question

**Core time:** 2 hours. **Prerequisites:** foundations and at least one pillar. **Outcomes:** delimit a research question; build a primary-source evidence map; separate reproduction, replication and extension; write a short protocol before inspecting final outcomes.

## 1. Reduce scope until the question has a measurable answer

“Make AI trustworthy” is a motivation, not a mini-project question. “Under a fixed local digit experiment, how does one training-poisoning fraction affect clean accuracy and non-target trigger ASR?” identifies an intervention, outcomes and setting. A small question can produce a rigorous answer within the onboarding period. A broad question with many moving parts often produces a collection of numbers that cannot support a conclusion.

Write the population, model, intervention, comparator, outcomes and budget. Identify which assumption, if false, would change your interpretation. A useful project can test a known mechanism in a new small setting; novelty claims require a literature review and should not be asserted merely because you have not seen the idea before.

## 2. Read papers as arguments

For each primary paper, record the problem, claimed contribution, formal assumptions, algorithm, evaluation data, baselines, metrics, computational scope and limitations. Separate what the authors measured from the explanation they propose and from your own hypothesis. Abstracts help orient you but usually omit details needed to reproduce an experiment.

Prefer the publisher/proceedings page, original paper and authors' implementation when available. Check title, author list, year, version and venue. An arXiv posting is not automatically a peer-reviewed publication; a later conference version may differ. Cite the version you actually used. Do not cite a secondary summary as though it were the original source of an algorithm.

| Evidence-map column | What to write |
|---|---|

| Claim | A narrow paraphrase in your own words |
| Assumptions | Data, threat, optimization or system conditions |
| Evidence | Experiments/theory actually supplied |
| Gap or uncertainty | What the source does not establish |
| Your use | Background, baseline, reproduced mechanism or extension |

## 3. Worked narrowing process

Start with “Federated learning is bad on non-IID data.” This is too universal. Narrow to a fixed model and partition family. Choose a global test population, local-step budget and aggregation rule. A final question might be: “For a four-feature synthetic binary problem with five clients, does increasing local full-gradient steps from 1 to 5 change the IID-versus-label-skew validation trajectory at matched communication rounds?” The question still has a budget tradeoff—local computation differs—which must be reported.

A hypothesis could predict larger client deltas under skew. The primary prediction metric might remain unchanged. That result can support a mechanism diagnostic without demonstrating a utility degradation. Define both quantities so you do not change the intended claim after seeing the curves.

## 4. Label the type of project accurately

This program uses “reproduction” for rerunning or independently implementing a specified published procedure, with exact meaning explained in the report because communities use the terms differently. A teaching demonstration illustrates an idea on a simplified setup. An extension changes a defined element beyond the reproduced baseline. Do not call a logistic CPU simulator a reproduction of a large neural-network system's numerical results.

Your report should contain a fidelity table: original setting, your setting, reason for each difference and expected implication. This turns resource constraints into transparent scope rather than hidden deviations.

## 5. Predeclare a protocol

Before final evaluation, save a dated protocol containing the question, hypotheses, primary/secondary metrics, split, tuning budget, seed schedule, stop rule and analysis plan. Commit it in version control or preserve another timestamped record. If you later change it, record the amendment and why. Exploratory work is legitimate; pretending it was predeclared is not.



```python
protocol = {
 'question':'Effect of local steps under a fixed label-skew partition',
 'factor':{'local_steps':[1,5]},
 'primary_metric':'held-out accuracy at fixed final round',
 'secondary_metrics':['client macro accuracy','calculated payload bytes'],
 'selection_data':'validation only',
 'seeds':[7,17,27],
 'stop_rule':'30 rounds for every configuration'
}
assert protocol['selection_data'] == 'validation only'
```



This schema check does not prove you followed the protocol. The run history and artifacts provide that evidence.

# Build an experiment package another researcher can inspect

**Core time:** 2 hours. **Prerequisites:** a focused question and reproducible lab experience. **Outcomes:** organize code/configuration/data provenance; preserve raw evidence; define checks and a clean reproduction command; document data and artifact boundaries.

## 1. Separate source, configuration and results

Put reusable implementation in source files, experimental choices in a configuration and generated outputs in a run-specific directory. A notebook can explain the analysis, but hidden execution state must not be required for reproduction. Restart and run all cells in order before submitting. Avoid manual spreadsheet edits that cannot be regenerated from raw outputs.

A small package can contain `README.md`, `requirements.txt`, `protocol.json`, `src/`, `configs/`, `results/`, `report.pdf` and `LICENSES.md`. Include the exact command for the main result and a smaller smoke command. Explain whether the command retrains, evaluates saved weights or only regenerates plots; these are different tasks.

## 2. Data provenance makes indices meaningful

Record source URL, version, license/usage conditions, download or generation method, preprocessing and split identifiers. For generated data, save generator parameters and seeds. For external data, retain attribution and verify what redistribution is permitted. If data cannot be shared, provide an authorized access path and a synthetic smoke example without pretending the latter reproduces the real-data result.

Do not place credentials, student records or confidential research data in the repository, notebook outputs or shared screenshots. The course's bundled/synthetic data avoid those needs. If your proposed extension requires restricted data, discuss the data-governance requirements with your supervisor before changing the scope.

## 3. Worked artifact identity check

A file checksum helps verify that a colleague has the same bytes. It does not prove that the file's contents are correct, legally usable or unmodified before the checksum was recorded. Store hashes together with a description of what each artifact represents.



```python
import hashlib
from pathlib import Path
def sha256(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as stream:
        for block in iter(lambda:stream.read(1024*1024),b''):
            digest.update(block)
    return digest.hexdigest()
# Call sha256 on your saved configuration, split file and final metrics file.
assert len(hashlib.sha256(b'SAIL teaching example').hexdigest()) == 64
```



A code commit ID identifies a version-controlled source state, while uncommitted local changes can still alter a run. Record whether the working tree was clean. Do not embed environment secrets in an “environment dump”; record package/device information selectively.

## 4. Checks should match the scientific claim

For a robustness project, verify attack bounds, eligibility and clean controls. For FL, verify partition disjointness, weighting and local initialization. For efficiency, verify equivalent inputs, representation labels and timing boundaries. For all projects, verify no test-driven selection, finite metrics and recoverable denominators. A generic “script exited zero” check is necessary but not sufficient.

Use `lab07_project.py --root YOUR_OUTPUT_FOLDER` to create a structural artifact inventory and hashes. Its report explicitly states that it does not validate scientific correctness, privacy or the absence of all secrets. You still need the method-specific checks and a reviewer.

## 5. Keep raw and derived evidence connected

Save per-run metrics and relevant predictions before computing means, intervals or plots. A figure caption should identify the configuration, sample size, aggregation and uncertainty definition. If a plot excludes failed runs, explain the exclusion rule and list them. Prefer a predeclared failure policy to decisions made after seeing whether the failure favors the hypothesis.

A run log should record start/end status, command, configuration, output path and error if failed. A failed experiment can reveal a feasibility limitation; erasing it removes evidence. Do not merge results from different code versions without preserving that distinction.

**Estimated project effort:** 12–20 hours after the Trustworthy AI module. **Prerequisites:** the clean/poisoned local experiment, metric eligibility and a written protocol. Complete this capstone only if assigned by your administrator; the other capstone briefs are alternatives.

## 1. Research question and scope

Investigate how one controlled change in a local backdoor experiment affects clean utility and triggered target behavior. A recommended core question is: “For a fixed small classifier and visible trigger, how does training poisoning fraction affect clean accuracy and non-target ASR?” Use public bundled or synthetic data and your own locally trained models. Do not test third-party services or private datasets.

State differences in architecture, dataset, trigger, training budget and selection. A precise account of those differences is required even if the qualitative phenomenon matches.

## 2. Required experimental design

Use a clean training/validation/test split with saved indices. Select a fixed target and visible trigger. Predeclare a small poisoning grid, for example 0%,5%,10% of all training examples, drawn from non-target training examples. Record exact integer counts and the eligibility pool. Keep the clean baseline and use the same trigger on it as a control.

Repeat across at least three training/poison-selection seeds when feasible, keeping the split fixed for the core comparison. If varying splits, analyze that separately. Use the same architecture and training/selection policy for clean and poisoned models. Tune only on validation. Test settings must be frozen before the final table is produced.

The default lab is a starting point, not the completed project. Add at least one controlled extension: another poison fraction, a trigger-location control, a targeted defense with explicit knowledge assumptions or a classwise failure analysis. Change one principal factor at a time.

## 3. Required measurements

For a defense extension, include the undefended baseline, clean utility, attack procedure, defender knowledge and additional computation. If the defense knows the trigger, do not describe it as detecting arbitrary unknown triggers. No result in this project establishes universal absence of backdoors or certified robustness.

## 4. Worked analysis plan

For each seed and fraction, save counts before computing percentages. Compute paired seed-level differences in clean accuracy and ASR relative to the appropriate baseline. Show individual seeds and a descriptive aggregate. With only three seeds, avoid overconfident distributional claims. Explain whether uncertainty reflects training variation, poisoning selection or finite test sampling.



```python
def non_target_asr(predictions,original_labels,target):
    import numpy as np
    predictions=np.asarray(predictions);original_labels=np.asarray(original_labels)
    eligible=original_labels!=target
    count=int(eligible.sum())
    if count==0: return {'successes':0,'eligible':0,'rate':None}
    successes=int((predictions[eligible]==target).sum())
    return {'successes':successes,'eligible':count,'rate':successes/count}
assert non_target_asr([0,0,1],[0,1,2],0)['rate'] == .5
```

## 3. Correctness gates before training

Verify normalized sample-count weighting with an exact toy case. Verify one full local step under full participation matches the pooled gradient update under the stated assumptions. Check that each client starts from a copy of the same round model. Check nonempty client sets, disjointness and coverage. Confirm mu=0 matches the nonproximal implementation with all other settings identical.



```python
import numpy as np
models=np.array([[1.,2.],[3.,0.],[5.,4.]])
counts=np.array([2,1,1])
server=np.average(models,axis=0,weights=counts)
np.testing.assert_allclose(server,[2.5,2.])
assert counts.sum()==4
print(server)
```



A high final score cannot substitute for these checks: an accidentally centralized or sequentially shared model can also classify well.

# Capstone brief: Efficient Model Trade-off Study

## 3. Worked storage gate

Before testing rank r for an m-by-n matrix, compare `r*(m+n)` with `m*n`. For a 64-by-64 matrix, rank 8 reduces 4,096 weight values to 1,024 factor values; rank 40 increases them to 5,120. This arithmetic filters unreasonable compression settings but does not predict latency or accuracy.



```python
def factor_storage(m,n,r):
    if not 1<=r<=min(m,n): raise ValueError('Invalid rank')
    return {'dense_values':m*n,'factor_values':r*(m+n),
            'saves_values':r*(m+n)<m*n}
assert factor_storage(64,64,8)['saves_values']
assert not factor_storage(64,64,40)['saves_values']
```



Verify the implemented factor shapes, bias placement and full-rank equivalence before interpreting a low-rank result. Similar mechanism-specific checks apply to masks, quantization ranges and frozen-teacher parameters.

## 4. Required measurements

Report accuracy, macro F1, class supports, parameter count, raw parameter bytes, actual serialized bytes and latency median/p95 for at least one declared batch size. Preserve raw timing samples and repeat complete sessions if differences are small. Include a Pareto plot for quality versus the primary resource and explain the decision rule.

Keep analytical metrics separate from measured metrics. Counted FLOPs are not measured energy; parameter bytes are not peak RAM; compressed disk bytes are not necessarily runtime storage. If the result depends on backend conversion, show which representation was executed.

## 5. Analysis and deliverables

# Write the report, reproduce the evidence and defend the conclusion

**Core time:** 2–3 hours, plus revision. **Prerequisites:** completed assigned experiments and artifact inventory. **Outcomes:** connect claims to evidence; write informative captions and limitations; reproduce a selected result; respond to review constructively.

## 1. Organize around the research argument

Begin with the problem, precise question and why the chosen comparison answers it. Describe data, assumptions, methods and selection before presenting results. Show the baseline and controlled intervention with matching metric definitions. Interpret what changed and what alternative explanations remain. End with a conclusion bounded by the tested setting and one justified next experiment.

Avoid narrating every command chronologically. A reader needs the final protocol and evidence, plus deviations that affect interpretation. Preserve the full run history in artifacts. If the scope changed, update the title and abstract so they describe the work actually completed rather than an abandoned ambition.

## 2. Every quantitative claim needs a traceable source

For each reported value, identify its run/configuration, raw file, metric formula, numerator/denominator or aggregation rule. A figure caption should state the population, number of seeds, error-bar definition and relevant budget. “Mean ± error” is incomplete. Say standard deviation across training seeds, a particular bootstrap interval or another defined quantity.

Do not quote tutorial numbers as your own measurements. Cite ideas and algorithms, then distinguish your implementation and results. Paraphrase source claims accurately and link the primary source. A citation does not make an unsupported extrapolation valid.

## 3. Worked conclusion revision

Weak conclusion: “FedProx solves non-IID learning.” Stronger conclusion: “In the specified five-client synthetic logistic experiment, the tested proximal setting reduced the recorded local-update norm, while final accuracy remained similar to FedAvg under the fixed round budget. The experiment does not establish benefits for neural models, unseen clients or networked deployment.” The stronger statement says what was measured, separates mechanism from utility and names the untested scope.

For efficiency, “the file shrank” should not become “the model uses less RAM and energy.” For trustworthiness, “the tested attack failed” should not become “the model is secure.” These are changes in claim, not stylistic improvements.

## 4. A report structure that fits this program

Use a 4–6 page main report for the capstone, with appendices for detailed configurations. Suggested sections are abstract/question, related primary work and fidelity, protocol/method, results, analysis/limitations and reproduction instructions. Include one main comparison table and figures that answer the question. Extra charts that repeat the same number do not add evidence.

The abstract should name the setting and principal observation without overstating novelty. The methods should be sufficient to reconstruct the comparison. The limitations should identify concrete boundaries, not merely “more data would help.” The artifact README should contain exact commands and expected output file names.

## 5. Reproduction rehearsal

Start from a fresh output directory and follow your own README. Verify dependencies, data provenance, split hashes and configuration. Regenerate one main table/figure from raw metrics, then rerun a small experiment or the full core experiment as feasible. Record any nondeterministic differences and whether they alter the conclusion. Do not demand exact timing equality across machines.

Use `lab07_project.py` to inventory artifact sizes and hashes. Inspect the report's limitations: structural completeness does not prove correctness. A peer should also challenge the metric formula, baseline fairness and whether the experiment actually tests the question.

## 6. Prepare a short oral defense

In 5–8 slides, explain the question, primary source, protocol, main result, one failure/limitation and next step. Be ready to derive one equation or recompute one table entry. If you do not know an answer, identify the missing evidence and propose a test. Inventing certainty is worse than a precise unresolved question.

Reviewer feedback may request a correction to code, a clearer denominator, a missing baseline or a narrower claim. Keep a revision log mapping each requested change to the edited artifact and new evidence. The portal permits revised submission only when Revision is requested and the deadline is open; reading completion alone does not determine the final result.
